## **C. CATEGORY-AVERAGE VISUALIZATION**

```python
df['Average'] = df[['Math', 'Electronics', 'GEAS', 'Communication']].mean(axis=1)
```

```python
Track = df.groupby('Track')[['Average']].mean().reset_index()
Track
```

```python
Gender = df.groupby('Gender')[['Average']].mean().reset_index()
Gender
```

```python
Hometown = df.groupby('Hometown')[['Average']].mean().reset_index()
Hometown
```

```python
import matplotlib.pyplot as plt
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5), sharey=True)

axes[0].bar(Track['Track'], Track['Average'])
axes[0].set_title('Mean Average by Track')
axes[0].set_ylabel('Mean Avreage')

axes[1].bar(Gender['Gender'], Gender['Average'], color='red')
axes[1].set_title('Mean Average by Gender')

axes[2].bar(Hometown['Hometown'], Hometown['Average'], color='yellow')
axes[2].set_title('Mean Average by Hometown')

plt.ylim(0, 100)
plt.tight_layout()
plt.show()
```

**Track** - The communication track has the highest recorded average score of 67.976.

**Gender** - Male students have the highest recorded average score of 67.183.

**Hometown** - Students originating from Luzon have the highest recorded average score of 68.083.
